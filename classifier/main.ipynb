```python
from baseline_quality_estimator import estimate_sharpness
import shutil
from pathlib import Path

SOURCE_DIRS = [
    Path("../database/albumentations"),
    Path("../database/augraphy"),
]

LABEL = Path("../database/labels/rg_real.json")
```

```python
LEGIBLE = Path("../database/sobel/legible")
LEGIBLE.mkdir(parents=True, exist_ok=True) 

ILLEGIBLE = Path("../database/sobel/illegible")
ILLEGIBLE.mkdir(parents=True, exist_ok=True)

THRESHOLD = 80
```

```python

for source in SOURCE_DIRS:

    for input_img_fn in source.glob("*"):

        score = estimate_sharpness(str(input_img_fn))

        output_file_name = ( LEGIBLE / input_img_fn.name ) if score > THRESHOLD else ( ILLEGIBLE / input_img_fn.name )

        print(f"{input_img_fn.name} -> {score:.2f} -> {output_file_name}")

        shutil.copy2(input_img_fn, output_file_name)
```
